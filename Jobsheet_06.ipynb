Praktikum 05 - Operasi pada List 2D

```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7, 8, 9], [10, 11, 12]]
C = []

for i in range(len(A)):
  baris = []
  for j in range(len(A[0])):
    baris.append(A[i][j] + B[i][j])
  C.append(baris)

print(C)
```

Penugasan 02 - Operasi Dasar List

```python
# Buat list nilai
nilai = [70, 75, 80]
print(f"List Nilai Awal: {nilai}")

# • Ubah nilai kedua (indeks 1) menjadi 85
nilai[1] = 85
print(f"Setelah Ubah Nilai Kedua: {nilai}")

# • Tambahkan nilai 90
nilai.append(90)
print(f"Setelah Tambah Nilai 90: {nilai}")

# • Hapus nilai 70
# Gunakan .remove() untuk menghapus berdasarkan nilai
nilai.remove(70)
print(f"Setelah Hapus Nilai 70: {nilai}")

# • Cetak hasil akhir
print(f"Hasil Akhir Operasi Dasar List: {nilai}")
```

Penugasan 03 - Slicing List

```python
# Buat list angka dari 1 sampai 10
angka = list(range(1, 11))
print(f"List Angka: {angka}")

# • 3 elemen pertama
tiga_pertama = angka[:3]
print(f"• 3 elemen pertama: {tiga_pertama}")

# • 3 elemen terakhir
empat_terakhir = angka[-4:]
print(f"• 4 elemen terakhir: {empat_terakhir}")

# • Elemen dengan indeks genap saja (indeks 0, 2 , 4, ...)
# Slicing dengan langkah (step) 2, dimulai dari indeks 0
indeks_genap = angka[1::2]
print(f"•Elemen dengan indeks genap saja: {indeks_genap}")
```

Penugasan 04 - List 2D - Matriks

```python
# Buat matriks 2x3 berisi angka 1-6
# Baris 1 = [1, 2, 3]
# Baris 2 = [4, 5, 6]
matriks = [
    [1, 2, 3],
    [4, 5, 6]
]
print("Matriks 2x3:")
for row in matriks:
  print(row)

# • Baris pertama kolom kedua (indeks [0][1])
baris_1_kolom_2 = matriks[0][1]
print(f"• Baris pertama kolom kedua: {baris_1_kolom_2}")

# • Baris kedua kolom ketiga (indeks [1][2])
baris_2_kolom_3 = matriks[1][2]
print(f"• Baris kedua kolom ketiga: {baris_2_kolom_3}")

# • Seluruh baris pertama (indeks [0])
seluruh_baris_pertama = matriks[0]
print(f"• Seluruh baris pertama: {seluruh_baris_pertama}")
```

Penugasan 05 - Operasi Matriks 2D

```python
# Buat dua matriks 2x2
matriks_a = [
    [1, 2],
    [3, 4]
]

matriks_b = [
    [5, 6],
    [7, 8]
]

# Inisialisasi matriks hasil penjumlahan dengan ukuran yang sama, diisi 0
hasil_penjumlahan = [
    [0, 0],
    [0, 0]
]

# Hitung hasil penjumlahan
# Loop untuk baris (i)
for i in range(len(matriks_a)):
  # Loop untuk kolom (j)
  for j in range(len(matriks_a[0])):
    hasil_penjumlahan[i][j] = matriks_a[i][j] + matriks_b[i][j]

print("Matriks A:")
for row in matriks_a:
  print(row)
print("\nMatriks B:")
for row in matriks_b:
  print(row)
print("\nHasil Penjumlahan Matriks (A + B):")
for row in hasil_penjumlahan:
  print(row)
```

Penugasan 06 - Rata-rata Nilai Mahasiswa

```python
# Data nilai mahasiswa (List 2D)
nilai = [[75, 80, 85], [60, 70, 65], [90, 95, 92]]
nama = ["Mahasiswa A", "Mahasiswa B", "Mahasiswa C"] #Asumsi nama

print("Data Nilai Mahasiswa:")
for i, row in enumerate(nilai):
  print(f"{nama[i]}: {row}")

print("\nRata-rata Nilai Setiap Mahasiswa:")

# Hitung dan tampilkan rata-rata nilai setiap mahasiswa
for i in range(len(nilai)):
  # Ambil list nilai untuk mahasiswa ke-i
  nilai_mahasiswa = nilai[i]

  # Hitung jumlah total nilai
  total_nilai = sum(nilai_mahasiswa)

  # Hitung rata-rata
  rata_rata = total_nilai / len(nilai_mahasiswa)

  # Tampilkan hasil
  print(f"• {nama[i]}: Total = {total_nilai}, Rata-rata = {rata_rata:.2f}")
```

Penugasan 07 - Studi Kasus - Data Toko

```python
# Data stok barang (List 2D: [Nama Barang, Stok])
stok = [["Buku", 20], ["Pulpen", 35], ["Penghapus", 15], ["Pensil", 40]]

# Variabel untuk menyimpan barang dengan stok terbanyak
stok_terbanyak = 0
barang_terbanyak = ""

# Program yang menampilkan daftar barang dan mencari stok terbanyak
print("Daftar Stok Barang Toko:")
print("-" * 30)
print(f"{'Nama Barang':<15}{'Stok':>10}")
print("-" * 30)

for item in stok:
  nama_barang = item[0]
  jumlah_stok = item[1]

  # Menampilkan daftar barang dan stoknya
  print(f"{nama_barang:<15}{jumlah_stok:>10}")

  # Mencari barang dengan stok terbanyak
  if jumlah_stok > stok_terbanyak:
      stok_terbanyak = jumlah_stok
      barang_terbanyak = nama_barang

print("-" * 30)
print(f"Barang dengan Stok Terbanyak adalah **{barang_terbanyak}** dengan jumlah **{stok_terbanyak}**.")
```
