Penugasan 01 - Eksperimen dengan Modul math

```python
import math

def luas_lingkaran(r):
  return math.pi * r * r

r1 = 7
r2 = 10

luas1 = luas_lingkaran(r1)
luas2 = luas_lingkaran(r2)

print("Luas lingkaran dengan jari-jari", r1, "adalah:", luas1)
print("Luas lingkaran dengan jari-jari", r2, "adalah:", luas2)
```

Penugasan 04 - Modul Buatan Sendiri

```python
%%writefile konversi.py

def c_to_f(angka):
  f = angka * 1.8 + 32
  print(f"Konversi dari {angka} Celcius ke Fahrenheit adalah: °{f}")

def f_to_c(angka):
  c = (angka - 32) / 1.8
  print(f"Konversi dari {angka} Fahrenheit ke Celcius adalah: °{c}")

```

```python
import konversi

print("-----Konversi-----")
print("1. Celcius ke Fahrenheit")
print("2. Fahrenheit ke Celcius")
print()
pilih = int(input("Masukkan angka: "))

if pilih == 1:
  angka = int(input("Masukkan angka yang ingin anda konversi:"))
  konversi.c_to_f(angka)

else:
  angka = int(input("Masukkan angka yang ingin anda konversi:"))
  konversi.f_to_c(angka)
```
