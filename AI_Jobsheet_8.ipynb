```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('iklan_sosmed.csv')
display(df.head())
```

## Data Preparation - One Hot Encoding

```python
encoder = OneHotEncoder(sparse_output=False) # Ubah jadi sparse=False jika library scikit-learn kamu versi lama
encoded_df = pd.DataFrame(encoder.fit_transform(df[['Jenis_Kelamin']]), columns=encoder.get_feature_names_out(['Jenis_Kelamin']))
df_encoded = pd.concat([df.drop(columns=['Jenis_Kelamin']), encoded_df], axis=1)
```

## Data Preparation - Outlier Removal

```python
# Mencari nilai z-score untuk setiap data point pada kolom Umur dan Gaji
z_scores_umur = np.abs((df_encoded['Umur'] - df_encoded['Umur'].mean()) / df_encoded['Umur'].std())
z_scores_gaji = np.abs((df_encoded['Gaji'] - df_encoded['Gaji'].mean()) / df_encoded['Gaji'].std())

# Menghilangkan data point yang memiliki z-score lebih besar dari 3 pada kedua kolom
df_clean = df_encoded[(z_scores_umur < 3) & (z_scores_gaji < 3)].copy()

# Menampilkan informasi data sebelum dan sesudah dibersihkan
print(f"Jumlah data sebelum outlier removal: {len(df_encoded)}")
print(f"Jumlah data setelah outlier removal: {len(df_clean)}")
```

## Data Preparation - Normalization

```python
# Membuat instance untuk normalisasi
scaler = MinMaxScaler()

# Melakukan normalisasi pada kolom Umur dan Gaji
df_clean[['Umur', 'Gaji']] = scaler.fit_transform(df_clean[['Umur', 'Gaji']])

# Menampilkan hasil data yang sudah dinormalisasi
df_clean.head()
```

## Data Preparation - Cross Validation

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier

# 1. Membuat instance model Decision Tree
dt = DecisionTreeClassifier(random_state=42)

# Define features (X) and target (y)
X = df_clean.drop('Transaksi', axis=1)
y = df_clean['Transaksi']

# 2. Melakukan 5-fold cross validation menggunakan langsung variabel X dan y
# (pastikan cell Langkah 5 sebelumnya sudah di-run agar X dan y terbaca)
scores = cross_val_score(dt, X, y, cv=5)

# 3. Menampilkan hasil cross validation
print('Hasil cross validation per fold:', scores)
print('Rata-rata akurasi model:', scores.mean())
```
