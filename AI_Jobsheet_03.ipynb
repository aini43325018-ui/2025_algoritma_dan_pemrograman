Latihan 02: Spread

```python
# Mendapatkan Nilai Range

import numpy as np

print("Mendapatkan Nilai Range")
data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]

minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum

print("Data :", data)
print("Minimum :", minimum)
print("Maximum:", maximum)
print("Range (x):", x)

# Mendapatlan Nilai Percentile
data = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.percentile(data, 90)

print(" Data :", data)
print("Percentile 90 ;", x)

# Mendapatkan Nilai Quartile dan Interquartile Range (IQR)
print("Mendapatkan Nilai Quartile dan IQR")
data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]
Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1

print("Data :", data)
print("Q1 :", Q1)
print("Q3 :",Q3)
print("IQR :", IQR)

# Mendapatkan Nilai Variance dan Standard Deviation")
print("Mendapatkan Nilai Variance dan Standard Deviation")
data = [86, 87, 88, 86, 87, 85, 86]
mean = np.mean(data)
var = np.var(data)
std = np.std(data)

print(" Data :", data)
print("Mean (Rata-rata) :", mean)
print("Variance :", var)
print("Standard Deviation :", std)
```

Praktikum 03

```python
import numpy as np

nilai = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]

# a. Urutkan data
nilai_urut = np.sort(nilai)
print("a. Data Terurut :", nilai_urut)

# b. Q1, Q3, dan IQR
Q1 = np.percentile(nilai_urut, 25)
Q3 = np.percentile(nilai_urut, 75)
IQR = Q3 - Q1
print(f"b. Q1: {Q1} | Q3: {Q3} | IQR: {IQR}")

# c. Upper Bound dan Lower Bound
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)
print(f"c. Lower Bound  : {lower_bound}")
print(f"c. Upper Bound  : {upper_bound}")

# d. Mencari Outlier (Data Pencilan)
outliers = [x for x in nilai_urut if x < lower_bound or x > upper_bound]
print("d. Outlier      :", outliers)
```

Praktikum 04

```python
import numpy as np

# Catatan: Asumsi interval kelas ke-3 adalah 51-55 agar seragam (panjang kelas = 5)
batas_bawah = np.array([41, 46, 51, 56, 61, 66])
batas_atas = np.array([45, 50, 55, 60, 65, 70])
frekuensi = np.array([2, 3, 6, 4, 5, 2])

# 1. Cari nilai tengah (Xi) tiap kelas
xi = (batas_bawah + batas_atas) / 2

# 2. Cari Rata-rata data kelompok
mean_kelompok = np.sum(frekuensi * xi) / np.sum(frekuensi)

# 3. Cari Varians & Standar Deviasi (Menggunakan rumus sampel n-1)
n = np.sum(frekuensi)
varians = np.sum(frekuensi * (xi - mean_kelompok)**2) / (n - 1)
std_deviasi = np.sqrt(varians)

print("Nilai Tengah (Xi)  :", xi)
print("Rata-rata (Mean)   :", mean_kelompok)
print("Standar Deviasi    :", std_deviasi)
```
