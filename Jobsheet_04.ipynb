Praktikum 01 - Perulangan For dengan Range

```python
for i in range(1, 5):
  print(i)
```

Praktikum 02 - Perulangan While

```python
n = int(input("Masukkan angka: "))
while n > 0:
  print(n)
  n -= 1
```

Praktikum 03 - Nested Loop (Segitiga Bintang)

```python
for i in range(1,5):
  for j in range(i):
    print("*", end="")
  print()
```

Praktikum 05 - For pada Dictionary

```python
data = {"Nama": "Andi", "Umur": 20, "Prodi": "TRK"}

for k, v in data.items():
  print(k, ":", v)
```

Praktikum 06 - For dengan Enumerate

```python
daftar = ["Andi", "Budi", "Cici"]

for i, nama in enumerate(daftar, start=1):
  print(i, nama)
```

Praktikum 07 - Studi Kasus: Program Kasir Multi-Item

```python


while True:
  barang = input("Masukkan nama barang (ketik 'selesai' untuk berhenti): ")
  if barang.lower() == "selesai":
    break
  harga = float(input(f"Masukkan harga {barang}: "))
  jumlah = int(input(f"Masukkan jumlah {barang}: "))
  subtotal = harga * jumlah

print("\n===== Struk Belanja =====")
print(f"Total Belanja : Rp {subtotal}")
```

Penugasan 02 - Hitung Faktorial

```python
n = int(input("Masukkan bilangan: "))
faktorial = 1
i = 1

while i <= n:
  faktorial *= i
  i += 1

print(f"Faktorial dari {n} adalah {faktorial}")
```

Penugasan 03 - Pola Segitiga Terbalik (Nested Loop)

```python
tinggi = 5

for i in range(tinggi, 0, -1):
  for j in range(i):
    print("*", end="")
  print()
```

Penugasan 04 - Cari Bilangan Prima

```python
bil = int(input("Masukkan bilangan:"))

if bil > 1:
  for i in range(2, bil):
    if bil % i -- 0:
      print(f"{bil} bukan bilangan prima")
      break
    else:
      print(f"{bil} adalah bilangan prima")
  else:
    print(f"{bil} bukan bilangan prima")
```

Penugasan 08 - Studi Kasus - Kasir Diskon

```python
total = 0

while True:
  nama = input("Masukkan nama barang (atau ketik 'selesai' untuk berhenti): ")
  if nama.lower() == "selesai":
    break
  harga = int(input(f"Harga {nama}: "))
  jumlah = int(input(f"Jumlah {nama}: "))
  subtotal = harga * jumlah
  total += subtotal
  print(f"Subtotal {nama}: Rp{subtotal}\n")

print("\n=== STRUK BELANJA ===")
print(f"Total Belanja: Rp{total}")


if total >= 200000:
  diskon = total * 0.10
  total -= diskon
  print(f"Diskon 10%: Rp{diskon}")
else:
  print("Diskon: Rp0")

print(f"Total yang harus dibayar: Rp{total}")
print("Terima kasih telah berbelanja!")
```
