Praktikum 01 - Bubble Sort

```python
def bubble_sort(data) :
    n = len(data)
    for i in range(n):
      for j in range(0, n - i - 1):
        if data[j] > data[j + 1]:
          data[j], data[j + 1] = data[j + 1], data[j]
    return data

angka = [64, 34, 25, 12, 22, 11, 90]
print("Data sebelum diurutkan :", angka)
print("Data setelah Bubble Sort :", bubble_sort(angka))
```

Praktikum 02 - Selection Sort

```python
def selection_sort(data):
  n = len(data)
  for i in range(n):
    min_index = i
    for j in range(i + 1, n):
      if data[j] < data[min_index]:
        min_index = j
    data[i], data[min_index] = data[min_index], data[i]
  return data

angka = [64, 25, 12, 22, 11]
print("Data sebelum diurutkan:", angka)
print("Data setelah Selection Sort:", selection_sort(angka))
```

Praktikum 03 - Insertion Sort

```python
def insertion_sort(data) :
  n = len(data)
  for i in range(1, n):
    key = data[i]
    j = i - 1
    while j >= 0 and data[j] > key:
      data[j + 1] = data[j]
      j -= 1
    data[j + 1] = key
  return data

angka = [12, 11, 13, 5, 6]
print("Data sebelum diurutkan:", angka)
print("Data setelah Insertion Sort:", insertion_sort(angka))
```

Praktikum 04 - Studi Kasus: Mengurutkan Nilai Mahasiswa

```python
def bubble_sort_nilai(data):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if data[j][1] < data[j + 1][1]:
        data[j], data[j + 1] = data[j + 1], data[j]
  return data

mahasiswa = [("Andi", 80), ("Budi", 70), ("Cici", 90), ("Doni", 85)]
print("Data sebelum diurutkan:", mahasiswa)
print("Data setelah diurutkan (desc):", bubble_sort_nilai(mahasiswa))
```

Praktikum 05 - Perbandingan Waktu Eksekusi Sorting

```python
import random, time

def bubble_sort(data):
    n = len(data)
    for i in range(n):
        for j in range(0, n - i - 1):
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
    return data

def selection_sort(data):
    n = len(data)
    for i in range(n):
        min_index = i
        for j in range(i + 1, n):
            if data[j] < data[min_index]:
                min_index = j
        data[i], data[min_index] = data[min_index], data[i]
    return data

def insertion_sort(data):
    n = len(data)
    for i in range(1, n):
        key = data[i]
        j = i - 1
        while j >= 0 and data[j] > key:
            data[j + 1] = data[j]
            j -= 1
        data[j + 1] = key
    return data

data = [random.randint(1, 1000) for _ in range(500)]
fungsi_sort = [bubble_sort, selection_sort, insertion_sort]

for func in fungsi_sort:
    salinan = data.copy()
    start = time.time()
    func(salinan)
    end = time.time()
    print(f"{func.__name__} membutuhkan waktu {end - start:.5f} detik")
```

Praktikum 06 - Sorting Data Barang (Nama dan Harga)

```python
def bubble_sort_key(data, key_index=0, reverse=False):
  n = len(data)
  for i in range(n):
    for j in range(0, n - i - 1):
      if reverse:
        if data[j][key_index] < data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
      else:
        if data[j][key_index] > data[j + 1][key_index]:
          data[j], data[j + 1] = data[j + 1], data[j]
    return data

barang = [("Buku", 15000), ("Pensil", 5000), ("Tas", 75000), ("Pulpen", 10000)]
print("Urut berdasarkan nama (asc):", bubble_sort_key(barang.copy(), 0))
print("Urut berdasarkan harga (desc):", bubble_sort_key(barang.copy(), 1, reverse=True))
```

Penugasan 01 - Eksperimen Bubble Sort

```python
def bubble_sort_desc(arr):
  n = len(arr)
  for i in range(n):
    for j in range(0, n - i - 1):
      if arr[j] < arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

data = [7, 2, 9, 3, 4]
print("Hasil descending :", bubble_sort_desc(data))
```

Penugasan 02 - Eksperimen Selection Sort

```python
def selection_sort_largest(arr):
  n = len(arr)
  for i in range(n - 1, 0, -1):
    max_index = 0
    for j in range(1, i + 1):
      if arr[j] > arr[max_index]:
        max_index = j
    arr[i], arr[max_index] = arr[max_index], arr[i]
  return arr

data = [12, 8, 97, 50, 25]
print("Urut (largest to end):", selection_sort_largest(data))
```

Penugasan 03 - Eksperimen Insertion Sort

```python
def insertion_sort_strings(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1
    while j >= 0 and arr[j].lower() > key.lower():
      arr[j + 1] = arr[j]
      j -= 1
    arr[j + 1] = key
  return arr

nama = ["Diana", "Nabila", "Syahda", "Mima"]
print("Sorted :", insertion_sort_strings(nama))
```

Penugasan 05 - Analisis Kompleksitas

```python
def bubble_sort_count(arr):
  count = 0
  n = len(arr)
  for i in range(n):
    for j in range(n - i - 1):
      count += 1
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return count

def selection_sort_count(arr):
  count = 0
  n = len(arr)
  for i in range(n):
    min_idx = i
    for j in range(i + 1, n):
      count += 1
      if arr[j] < arr[min_idx]:
        min_idx = j
    arr[i], arr[min_idx] = arr[min_idx], arr[i]
  return count

def insertion_sort_count(arr):
  count = 0
  for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1
    while j >= 0:
      count += 1
      if arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1
      else:
        break
    arr[j + 1] = key
  return count

import random

for n in [10, 70, 100]:
  data = [random.randint(1, 1000) for _ in range (n)]
  print("\nUkuran :", n)
  print("Bubble :", bubble_sort_count(data.copy()))
  print("Selection :", selection_sort_count(data.copy()))
  print("Insertion :", insertion_sort_count(data.copy()))
```

Penugasan 06 - Proyek Mini

```python
def bubble_sort(arr, key, reverse=False):
  n = len(arr)
  for i in range(n):
    for j in range(n - i - 1):
      if reverse:
        if arr[j][key] < arr [j + 1][key]:
          arr[j], arr[j + 1] = arr[j + 1 ], arr[j]
      else:
        if arr[j][key] > arr[j + 1][key]:
          arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr

barang = []

n = int(input("Jumlah Barang : "))
for _ in range(n):
  nama = input("Nama Barang : ")
  harga = int(input("Harga Barang :" ))
  barang.append({"nama": nama, "harga": harga})

print("\nPilihan Sort")
print("1. Nama Ascending")
print("2. Nama Descending")
print("3. Harga Ascending")
print("4. Harga Descending")

pilih = int(input("Pilih : "))

if pilih == 1:
  hasil = bubble_sort(barang, key="nama", reverse=False)
elif pilih == 2:
  hasil = bubble_sort(barang, key="nama", reverse=True)
elif pilih == 3:
  hasil = bubble_sort(barang, key="harga", reverse=False)
elif pilih == 4:
  hasil = bubble_sort(barang, key="harga", reverse=True)
else:
  print("Pilihan tidak valid")
  exit()

print("nHasil Sorting")
for b in hasil:
  print(b["nama"], "-", b["harga"])

```
