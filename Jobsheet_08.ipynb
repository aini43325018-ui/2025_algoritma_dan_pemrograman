Praktikum 04 - Fungsi dengan *args dan **kwargs

```python
def jumlahkan(*args):
  return sum(args)

def biodata(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")

print("Total:", jumlahkan(1, 2, 3, 4, 5))
biodata(Nama="Budi", Umur=20, Prodi="TRK")
```

Praktikum 07 - Fungsi Decorator

```python
def decorator_sapa(fungsi):
  def wrapper():
    print("Sebelum fungsi dijalankan...")
    fungsi()
    print("Setelah fungsi dijalankan...")
  return wrapper

@decorator_sapa
def sapa():
  print("Halo, Mahasiswa Polines!")

sapa()
```

Penugasan 07 - Fungsi Decorator

```python
def decorator_halo(fungsi):
  def logger():
    print("Memulai fungsi...")
    fungsi()
    print("Fungsi selesai dijalankan.")
  return logger

@decorator_halo
def halo():
  print("Halo, Zayn Malik")

halo()
```

Penugasan 08 - Studi Kasus Nyata - Program Nilai

```python
def input_nilai():
  print("=== Input Nilai Mahasiswa ===")
  n = int(input("Masukkan jumlah mahasiswa: "))
  nilai_mahasiswa = []
  for i in range(n):
    print(f"\nMahasiswa ke-{i+1}")
    tugas = float(input("Nilai Tugas: "))
    ujian = float(input("Nilai Ujian: "))

    # Lambda untuk menghitung nilai akhir (40% tugas, 60% ujian)
    nilai_akhir = (lambda t, u: (0.4 * t) + (0.6 * u))(tugas, ujian)
    nilai_mahasiswa.append(nilai_akhir)

  return nilai_mahasiswa

def hitung_rata(nilai_mahasiswa):
  return sum(nilai_mahasiswa) / len(nilai_mahasiswa)

def status_kelulusan(rata):
  if rata >= 60:
    return "LULUS"
  else:
    return "TIDAK LULUS"\

nilai = input_nilai()
rata_rata = hitung_rata(nilai)
status = status_kelulusan(rata_rata)

print("\n=== HASIL AKHIR ===")
print("Nilai Akhir Mahasiswa :", nilai)
print("Rata-rata Nilai       :", round(rata_rata, 2))
print("Status Kelulusan      :", status)
```
