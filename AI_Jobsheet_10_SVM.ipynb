## Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix
```

## Load Dataset

```python
# Membaca dataset
df_diabetes = pd.read_csv('diabetes.csv')
print("Dataset Diabetes berhasil dimuat!")
```

## Menampilkan Properti Data

```python
print("--- 5 Baris Pertama ---")
display(df_diabetes.head())

print("\n--- Informasi Dataset ---")
df_diabetes.info()
```

## Cek Missing Value dan Duplikasi Data

```python
print("--- Jumlah Missing Value ---")
print(df_diabetes.isnull().sum())

print(f"\nJumlah data duplikat: {df_diabetes.duplicated().sum()} baris")

# Menghapus data duplikat jika ada
df_diabetes = df_diabetes.drop_duplicates()
```

## Cek Korelasi dengan kolom Transaksi

```python
# Cek korelasi antar fitur
plt.figure(figsize=(10, 8))
sns.heatmap(df_diabetes.corr(), annot=True, cmap='Blues', fmt=".2f")
plt.title("Korelasi Antar Fitur dengan Target (Outcome)")
plt.show()

print("\nKorelasi spesifik terhadap kolom 'Outcome':")
print(df_diabetes.corr()['Outcome'].sort_values(ascending=False))
```

## Train Test Split

```python
# Pisahkan atribut (X) dan label (y)
X_svm = df_diabetes.drop('Outcome', axis=1)
y_svm = df_diabetes['Outcome']
```

```python
# Split data (33% testing sesuai modul)
X_train_svm, X_test_svm, y_train_svm, y_test_svm = train_test_split(X_svm, y_svm, test_size=0.33, random_state=42)
```

```python
# Feature Scaling (Wajib untuk algoritma SVM)
scaler_svm = StandardScaler()
X_train_svm = scaler_svm.fit_transform(X_train_svm)
X_test_svm = scaler_svm.transform(X_test_svm)

print("Ukuran X_train:", X_train_svm.shape)
print("Ukuran X_test:", X_test_svm.shape)
```

## Melatih Model SVM

```python
# Membuat objek SVC dan melatih model
svm_model = SVC(random_state=42)
svm_model.fit(X_train_svm, y_train_svm)
print("Model SVM selesai dilatih!")
```

## Evaluasi Data Test

```python
# Prediksi data test
y_pred_svm = svm_model.predict(X_test_svm)

# Menghitung akurasi
acc_svm = accuracy_score(y_test_svm, y_pred_svm)
print(f"Akurasi Model SVM: {acc_svm * 100:.2f}%")

# Menampilkan Confusion Matrix
print("\nConfusion Matrix:")
cm_svm = confusion_matrix(y_test_svm, y_pred_svm)
sns.heatmap(cm_svm, annot=True, fmt="d", cmap="Greens")
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.title('Confusion Matrix - SVM')
plt.show()
```

## SVR

## Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
```

## Load Dataset

```python
# Membaca dataset dengan menambahkan parameter delimiter
df_gaji = pd.read_csv('Data_Gaji.csv', delimiter=';')
print("Dataset Gaji berhasil dimuat dengan format yang benar!")
```

## Menampilkan Properti Data

```python
print("--- 5 Baris Pertama ---")
display(df_gaji.head())

print("\n--- Informasi Dataset ---")
df_gaji.info()
```

## Cek Missing Value dan Duplikasi Data

```python
print("--- Jumlah Missing Value ---")
print(df_gaji.isnull().sum())

print(f"\nJumlah data duplikat: {df_gaji.duplicated().sum()} baris")

df_gaji = df_gaji.drop_duplicates()
```

## Cek Korelasi dengan Kolom Target

```python
korelasi_gaji = df_gaji.corr()
print("Korelasi antara Tahun Pengalaman dan Gaji:")
display(korelasi_gaji)
```

## Menyiapkan Fitur (Train & Standarisasi)

```python
# Memisahkan atribut dan label
X_svr = df_gaji['Tahun_Pengalaman'].values.reshape(-1, 1) # Ubah jadi array 2D
y_svr = df_gaji['Gaji'].values

# Standarisasi data (Penting agar SVR bisa membuat garis yang akurat)
sc_X = StandardScaler()
sc_y = StandardScaler()

X_svr_scaled = sc_X.fit_transform(X_svr)
y_svr_scaled = sc_y.fit_transform(y_svr.reshape(-1, 1)).flatten()

print("Data siap dan sudah distandarisasi!")
```

## Melatih Model SVR

```python
# Membangun model dengan parameter C=1000, gamma=0.05, kernel='rbf' sesuai modul
svr_model = SVR(C=1000, gamma=0.05, kernel='rbf')

# Melatih model
svr_model.fit(X_svr_scaled, y_svr_scaled)
print("Model SVR selesai dilatih!")
```

## Melakukan Visualisasi Data SVR

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_svr_scaled, y_svr_scaled, color='blue', label='Data Asli')

# Membuat garis prediksi yang lebih mulus (mengurutkan nilai X)
X_grid = np.arange(min(X_svr_scaled), max(X_svr_scaled), 0.01).reshape(-1, 1)
plt.plot(X_grid, svr_model.predict(X_grid), color='red', linewidth=2, label='Garis Prediksi SVR')

plt.title('Support Vector Regression (SVR)')
plt.xlabel('Tahun Pengalaman (Scaled)')
plt.ylabel('Gaji (Scaled)')
plt.legend()
plt.show()
```

## Hyperparameter Tuning SVM

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load Data & Persiapan
df_diabetes = pd.read_csv('diabetes.csv')
X_svm = df_diabetes.drop('Outcome', axis=1)
y_svm = df_diabetes['Outcome']

# 2. Split & Standarisasi (Penting untuk SVM)
X_train_svm, X_test_svm, y_train_svm, y_test_svm = train_test_split(X_svm, y_svm, test_size=0.33, random_state=42)
scaler_svm = StandardScaler()
X_train_svm = scaler_svm.fit_transform(X_train_svm)
X_test_svm = scaler_svm.transform(X_test_svm)

# 3. Menerapkan Grid Search
print("Mencari parameter terbaik untuk SVM...")
param_grid_svm = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf', 'linear', 'poly']
}

grid_svm = GridSearchCV(SVC(random_state=42), param_grid_svm, refit=True, verbose=1, cv=5)
grid_svm.fit(X_train_svm, y_train_svm)

# 4. Hasil Tuning
print("\n=== HASIL TUNING SVM ===")
print("Parameter Terbaik:", grid_svm.best_params_)
y_pred_svm = grid_svm.predict(X_test_svm)
print(f"Akurasi SVM setelah Tuning: {accuracy_score(y_test_svm, y_pred_svm) * 100:.2f}%")
```

## Hyperparameter Tuning SVR

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler

# 1. Load Data & Persiapan
df_gaji = pd.read_csv('Data_Gaji.csv', delimiter=';')
X_svr = df_gaji['Tahun_Pengalaman'].values.reshape(-1, 1)
y_svr = df_gaji['Gaji'].values

# 2. Standarisasi (SVR sangat sensitif terhadap skala)
sc_X = StandardScaler()
sc_y = StandardScaler()
X_svr_scaled = sc_X.fit_transform(X_svr)
y_svr_scaled = sc_y.fit_transform(y_svr.reshape(-1, 1)).flatten()

# 3. Menerapkan Grid Search
print("Mencari parameter terbaik untuk SVR...")
param_grid_svr = {
    'C': [0.1, 1, 10, 100, 1000],
    'gamma': [1, 0.1, 0.01, 0.001, 'scale'],
    'kernel': ['rbf', 'linear', 'poly']
}

grid_svr = GridSearchCV(SVR(), param_grid_svr, refit=True, verbose=1, cv=5)
grid_svr.fit(X_svr_scaled, y_svr_scaled)

# 4. Hasil Tuning
print("\n=== HASIL TUNING SVR ===")
print("Parameter Terbaik:", grid_svr.best_params_)
print(f"Skor Akurasi (R-squared) SVR Terbaik: {grid_svr.best_score_:.4f}")

# 5. Visualisasi Hasil Terbaik SVR
plt.figure(figsize=(8,5))
plt.scatter(X_svr_scaled, y_svr_scaled, color='blue', label='Data Asli')
X_grid = np.arange(min(X_svr_scaled), max(X_svr_scaled), 0.01).reshape(-1, 1)
plt.plot(X_grid, grid_svr.predict(X_grid), color='red', linewidth=2, label='Garis Prediksi SVR Terbaik')
plt.title('SVR - Prediksi Gaji vs Tahun Pengalaman (Hasil Tuning)')
plt.xlabel('Tahun Pengalaman (Scaled)')
plt.ylabel('Gaji (Scaled)')
plt.legend()
plt.show()
```
